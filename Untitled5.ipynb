```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from datetime import datetime
import os
from pathlib import Path

# Create necessary directories
output_dir = Path('output')
viz_dir = output_dir / 'visualizations'
report_dir = output_dir / 'reports'

for dir in [output_dir, viz_dir, report_dir]:
    dir.mkdir(parents=True, exist_ok=True)

class LeadAnalyzer:
    def __init__(self, file_path):
        self.df = pd.read_csv(file_path)
        self.clean_data()

    def clean_data(self):
        # Convert dates
        self.df['lead_date'] = pd.to_datetime(self.df['lead_date'], format='%d-%m-%Y %H:%M')
        self.df['paid_at'] = pd.to_datetime(self.df['paid_at'], format='%d-%m-%Y %H:%M', errors='coerce')

        # Clean lead_id
        self.df['lead_id'] = self.df['lead_id'].str.replace(',', '').astype(float)

        # Convert graduation_year
        self.df['graduation_year'] = pd.to_numeric(self.df['graduation_year'])

        # Add derived columns
        self.df['is_converted'] = self.df['amount_paid'].notna()
        self.df['hour'] = self.df['lead_date'].dt.hour
        self.df['day_of_week'] = self.df['lead_date'].dt.day_name()

    def generate_channel_insights(self):
        channel_stats = pd.DataFrame({
            'total_leads': self.df.groupby('Channel_group').size(),
            'conversions': self.df[self.df['is_converted']].groupby('Channel_group').size(),
            'revenue': self.df.groupby('Channel_group')['amount_paid'].sum()
        })

        channel_stats['conversion_rate'] = (
            channel_stats['conversions'] / channel_stats['total_leads'] * 100
        ).round(2)

        return channel_stats

    def analyze_temporal_patterns(self):
        hourly_pattern = self.df.groupby('hour').size()
        daily_pattern = self.df.groupby('day_of_week').size()

        return hourly_pattern, daily_pattern

    def course_analysis(self):
        course_stats = pd.DataFrame({
            'total_leads': self.df.groupby('course').size(),
            'conversions': self.df[self.df['is_converted']].groupby('course').size(),
            'revenue': self.df.groupby('course')['amount_paid'].sum()
        })

        course_stats['conversion_rate'] = (
            course_stats['conversions'] / course_stats['total_leads'] * 100
        ).round(2)

        return course_stats

    def lead_type_analysis(self):
        lead_type_stats = pd.DataFrame({
            'total_leads': self.df.groupby('Lead_type').size(),
            'conversions': self.df[self.df['is_converted']].groupby('Lead_type').size()
        })

        lead_type_stats['conversion_rate'] = (
            lead_type_stats['conversions'] / lead_type_stats['total_leads'] * 100
        ).round(2)

        return lead_type_stats

    def create_visualizations(self):
        # 1. Channel Distribution and Conversion
        plt.figure(figsize=(15, 8))
        channel_stats = self.generate_channel_insights()

        ax1 = plt.gca()
        ax2 = ax1.twinx()

        channel_stats['total_leads'].plot(kind='bar', ax=ax1, color='skyblue')
        channel_stats['conversion_rate'].plot(kind='line', marker='o', ax=ax2, color='red')

        plt.title('Channel Performance: Lead Volume vs Conversion Rate')
        ax1.set_xlabel('Channel')
        ax1.set_ylabel('Total Leads')
        ax2.set_ylabel('Conversion Rate (%)')
        plt.tight_layout()
        plt.savefig(viz_dir / 'channel_performance.png')
        plt.close()

        # 2. Temporal Patterns
        hourly_pattern, _ = self.analyze_temporal_patterns()
        plt.figure(figsize=(12, 6))
        hourly_pattern.plot(kind='line', marker='o')
        plt.title('Lead Generation by Hour of Day')
        plt.xlabel('Hour')
        plt.ylabel('Number of Leads')
        plt.tight_layout()
        plt.savefig(viz_dir / 'hourly_pattern.png')
        plt.close()

        # 3. Course Performance
        course_stats = self.course_analysis()
        plt.figure(figsize=(12, 6))
        course_stats['conversion_rate'].plot(kind='bar')
        plt.title('Course-wise Conversion Rates')
        plt.xlabel('Course')
        plt.ylabel('Conversion Rate (%)')
        plt.xticks(rotation=45)
        plt.tight_layout()
        plt.savefig(viz_dir / 'course_performance.png')
        plt.close()

        # 4. Lead Type Analysis
        lead_type_stats = self.lead_type_analysis()
        plt.figure(figsize=(10, 6))
        lead_type_stats['conversion_rate'].plot(kind='bar')
        plt.title('Lead Type Conversion Rates')
        plt.xlabel('Lead Type')
        plt.ylabel('Conversion Rate (%)')
        plt.tight_layout()
        plt.savefig(viz_dir / 'lead_type_performance.png')
        plt.close()

    def generate_summary_stats(self):
        return {
            'total_leads': len(self.df),
            'total_conversions': self.df['is_converted'].sum(),
            'overall_conversion_rate': (self.df['is_converted'].mean() * 100).round(2),
            'total_revenue': self.df['amount_paid'].sum(),
            'top_performing_channel': self.generate_channel_insights().sort_values('conversion_rate').index[-1],
            'most_popular_course': self.df['course'].mode()[0],
            'best_converting_lead_type': self.lead_type_analysis().sort_values('conversion_rate').index[-1]
        }

def main():
    analyzer = LeadAnalyzer(fn)

    # Generate all analyses
    channel_insights = analyzer.generate_channel_insights()
    hourly_pattern, daily_pattern = analyzer.analyze_temporal_patterns()
    course_stats = analyzer.course_analysis()
    lead_type_stats = analyzer.lead_type_analysis()
    summary_stats = analyzer.generate_summary_stats()

    # Create visualizations
    analyzer.create_visualizations()

    # Generate detailed report
    generate_detailed_report(
        summary_stats,
        channel_insights,
        course_stats,
        lead_type_stats
    )

def generate_detailed_report(summary_stats, channel_insights, course_stats, lead_type_stats):
    report_template = """
# Lead Generation and Conversion Analysis Report
## Executive Summary

During the analysis period, we observed the following key metrics:
- Total Leads: {total_leads:,}
- Overall Conversion Rate: {conversion_rate:.2f}%
- Total Revenue: ${revenue:,.2f}
- Top Performing Channel: {top_channel}
- Most Popular Course: {top_course}

## Detailed Analysis

### 1. Channel Performance
{channel_performance}

### 2. Course Performance
{course_performance}

### 3. Lead Type Analysis
{lead_type_performance}

## Recommendations

1. **Channel Optimization**
   - Focus resources on {top_channel} given its superior performance
   - Implement targeted improvements for lower-performing channels
   - Develop channel-specific conversion strategies

2. **Course Marketing**
   - Leverage insights from {top_course}'s success
   - Develop specialized marketing approaches for each course
   - Consider course bundling opportunities

3. **Lead Type Strategy**
   - Prioritize {best_lead_type} leads for immediate follow-up
   - Develop targeted nurturing programs for each lead type
   - Implement lead scoring based on type performance

## Methodology and Assumptions

- Analysis period: Full dataset duration
- Conversion defined as completed payment
- All monetary values in USD
- Lead sources assumed to be consistently tracked
    """

    report_content = report_template.format(
        total_leads=summary_stats['total_leads'],
        conversion_rate=summary_stats['overall_conversion_rate'],
        revenue=summary_stats['total_revenue'],
        top_channel=summary_stats['top_performing_channel'],
        top_course=summary_stats['most_popular_course'],
        best_lead_type=summary_stats['best_converting_lead_type'],
        channel_performance=channel_insights.to_string(),
        course_performance=course_stats.to_string(),
        lead_type_performance=lead_type_stats.to_string()
    )

    # Save as markdown
    with open(report_dir / 'detailed_analysis.md', 'w') as f:
        f.write(report_content)

    # Convert to PDF (requires pypandoc)
    try:
        import pypandoc
        pypandoc.convert_file(
            str(report_dir / 'detailed_analysis.md'),
            'pdf',
            outputfile=str(report_dir / 'detailed_analysis.pdf'),
            extra_args=['--pdf-engine=xelatex']
        )
    except Exception as e:
        print(f"Could not generate PDF: {e}")
        print("Please install pypandoc and a LaTeX distribution to generate PDF")

if __name__ == "__main__":
    main()
```
